```python
df_general = pd.read_csv("Citas_Digital_General.csv")
```

```python
df_general["Fecha"] = pd.to_datetime(df_general["Fecha"], errors="coerce")
```

```python
meses = pd.date_range(start="2025-04-01", end="2026-07-01", freq="MS")
```

```python
for mes in meses:
    
    # Fecha inicial del mes
    fecha_inicio = mes
    
    # Fecha inicial del siguiente mes
    fecha_fin = mes + pd.DateOffset(months=1)
    
    # Filtramos las filas correspondientes al mes
    filtro_mes = df_general[
        (df_general["Fecha"] >= fecha_inicio) &
        (df_general["Fecha"] < fecha_fin)
    ]

     # Nombre del mes en español
    meses_espanol = {
        1: "Enero",
        2: "Febrero",
        3: "Marzo",
        4: "Abril",
        5: "Mayo",
        6: "Junio",
        7: "Julio",
        8: "Agosto",
        9: "Septiembre",
        10: "Octubre",
        11: "Noviembre",
        12: "Diciembre"
    }
    
    nombre_mes = meses_espanol[mes.month]
```

```python
  nombre_archivo = f"{nombre_mes}_{mes.year}.csv"
```

```python
filtro_mes.to_csv(nombre_archivo, index=False)
```
